## Recommended evaluation order

|Strategy	| Purpose |
| :---     | :---:    |
|Current dense baseline|	Existing reference point|
|Dense + reranker|	Already implemented|
|Sparse BM25 only|	Measures exact-term retrieval|
|Dense + sparse with RRF|	Tests hybrid complementarity|
|Hybrid + reranker|	Likely strongest retrieval pipeline|

Measure at least:

- hit@1
- recall@k
- MRR
- nDCG@k if graded relevance is available
- Retrieval latency
- End-to-end answer quality
- Citation correctness
- Abstention quality
